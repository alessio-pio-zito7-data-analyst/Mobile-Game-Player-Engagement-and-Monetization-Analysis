# Data Validation & Cleaning


### Importing pandas and load cvs file 

```python
import pandas as pd 

games = pd.read_csv('mobile_game_inapp_purchases.csv')

games
```

### Counting NaNs

```python
games.isna().sum()
```

136 player records contain missing values across all purchase-related fields (InAppPurchaseAmount, PaymentMethod, FirstPurchaseDaysAfterInstall and LastPurchaseDate), while still being assigned a SpendingSegment.

Since the purchase information cannot be verified, these records will be excluded from analyses involving revenue and monetization metrics.

They were retained for gameplay and engagement analyses, as gameplay-related fields (SessionCount and AverageSessionLength) remain complete.

```python
games[
(games['SpendingSegment'].isin(['Whale', 'Minnow', 'Dolphin'])) &
(games['InAppPurchaseAmount'].isna())]
```

```python
missing_purchase_record = games.loc[
    games['InAppPurchaseAmount'].isna(),
    'SpendingSegment'
].value_counts().reset_index().rename(columns={'count': 'MissingPurchaseRecords'})

missing_purchase_record
```

# Data Type conversion 

```python
games 
```

```python
games.dtypes
```

```python
df = games.copy() # a copy of our DataFrame has been created
```

```python
df['LastPurchaseDate'] = pd.to_datetime(df['LastPurchaseDate'], format = '%Y-%m-%d') 
df['Age'] = df['Age'].astype('Int64')
df['FirstPurchaseDaysAfterInstall'] = df['FirstPurchaseDaysAfterInstall'].astype('Int64') # 'Int64' preserves the NaN
df['GameGenre'] = df['GameGenre'].astype('category')
```

```python
df.info()
```

### Converting appropriate columns to optimized data types reduced memory usage and improved overall efficiency.
